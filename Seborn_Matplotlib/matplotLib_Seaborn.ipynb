```python
# Count Plot
# Used to count how many observations belong to each category.
import seaborn as sns
import matplotlib.pyplot as plt

data = ["Male", "Female", "Male", "Male", "Female"]

sns.countplot(x=data)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
print(df)
sns.countplot(
    data=df,
    x="class",
    hue="survived"
)

plt.show()
```

```python
# Histogram
# Used to understand the distribution of numerical data.
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90]

sns.histplot(marks, bins=5)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.histplot(
    data=df,
    x="sepal_length",
    hue="species",
    kde=True
)

plt.show()
```

```python
# Box Plot
# Very important for data analysis.
import seaborn as sns
import matplotlib.pyplot as plt

marks = [40, 45, 50, 55, 60, 65, 70, 100]

sns.boxplot(y=marks)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.boxplot(
    data=df,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
```

```python
# Violin Plot
# A violin plot combines ideas from a box plot + distribution plot
import seaborn as sns
import matplotlib.pyplot as plt

marks = [40, 45, 50, 55, 60, 60, 65, 70, 75, 80]

sns.violinplot(y=marks)

plt.show()
```

```python
# Boxen Plot
# Useful for large datasets because it shows more information about the distribution
import seaborn as sns
import matplotlib.pyplot as plt

data = [10, 12, 15, 20, 25, 30, 35, 40, 50]

sns.boxenplot(y=data)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("diamonds")

sns.boxenplot(
    data=df,
    x="cut",
    y="price"
)

plt.show()
```

```python
# Heatmap
# Very important in Machine Learning.
# It is commonly used to visualize a correlation matrix
import seaborn as sns
import matplotlib.pyplot as plt

data = [
    [1, 0.8, 0.3],
    [0.8, 1, 0.5],
    [0.3, 0.5, 1]
]

sns.heatmap(data, annot=True)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

correlation = df.select_dtypes("number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

```python
# Pair Plot
# Very useful when you want to see relationships between many numerical columns at once
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(df)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(
    df,
    hue="species",
    diag_kind="kde"
)

plt.show()
```

```python
# Joint Plot
# Shows the relationship between two variables plus their individual distributions.
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.jointplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.jointplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex",
    kind="scatter"
)

plt.show()
```

```python
# KDE Plot
# KDE = Kernel Density Estimate.
# It shows a smooth estimate of the distribution of data.
import seaborn as sns
import matplotlib.pyplot as plt

marks = [40, 45, 50, 55, 60, 60, 65, 70, 75, 80]

sns.kdeplot(marks)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.kdeplot(
    data=df,
    x="sepal_length",
    hue="species",
    fill=True
)

plt.show()
```

```python
# Regression Plot
# Used to show a relationship between variables with a regression line
# Very useful for understanding linear relationships.
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 8, 10]

sns.regplot(x=x, y=y)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.regplot(
    data=df,
    x="total_bill",
    y="tip"
)

plt.show()
# The line represents the estimated linear relationship.
```

```python
# lmplot()
# lmplot() is also used for linear regression, but it works especially well when you want to create multiple regression plots based on categories.
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.lmplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
```

```python
# . Strip Plot
# Shows individual data points for different categories.
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.stripplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
sns.stripplot(
    data=df,
    x="day",
    y="total_bill",
    hue="sex",
    jitter=True   #jitter=True separates overlapping points.
)

plt.show()
```
