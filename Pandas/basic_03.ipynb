```python
# Matplotlib is a popular Python library used for data visualization. It helps us create graphs and charts from data.

import pandas as pd
import matplotlib.pyplot as plt

# Our sample dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'Salary': [50000, 60000, 75000, 80000, 95000],
    'Bonus': [5000, 6000, 7500, 8000, 9500]
}
df = pd.DataFrame(data)
```

```python
# 1. Bar Chart
df.plot(kind='bar',x='Name',y='Salary',color='green')
plt.title("Employee Salaries")
plt.ylabel("Salary in USD")
plt.show()
```

```python
# 2. Pie Chart
df.set_index('Name')['Bonus'].plot(kind='pie',autopct='%1.1f%%',colormap='Pastel1')
plt.ylabel("") # Hides the default y-label for a cleaner look
plt.show()
```

```python
# 3. Line Plot
df.plot(kind='line', x='Age', y='Salary', marker='o', color='green')
plt.show()
```

```python
# 4. Histogram
import pandas as pd
import matplotlib.pyplot as plt
# Let's create a larger list of ages for a better histogram
ages = pd.DataFrame({'Employee_Ages': [22, 23, 24, 24, 25, 29, 30, 31, 35, 40, 42, 45, 50]})

# bins=5 means we want 5 buckets (e.g., 20-25, 26-30, etc.)
ages['Employee_Ages'].plot(kind='hist',bins=5, color='orange', edgecolor='black')

plt.title("Age Distribution in the Company")
plt.xlabel("Age Ranges")
plt.show()
```

```python
# 5. Correlation Heatmap
# 1. First, calculate the math (dropping text columns first)
numeric_df = df.drop('Name', axis=1)
correlation_matrix = numeric_df.corr()

# 2. Draw the heatmap using Matplotlib's matshow (Matrix Show)
plt.matshow(correlation_matrix, cmap='coolwarm')

# 3. Add a color legend on the side
plt.colorbar()

# 4. Label the axes so we know what we are looking at
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Heatmap", pad=20)
plt.show()
```

```python
# Create DataFrame
df = pd.DataFrame({
 "Maths": [80, 85, 90, 75, 85],
 "Science": [78, 82, 88, 80, 82],
 "English": [70, 75, 80, 72, 75]
})
print("DataFrame:")
print(df)
print("\nSUM:")
print(df.sum())
print(df.sum(axis=1))
print("\nMEAN:")
print(df.mean())
print(df.mean(axis=1))
print("\nMEDIAN:")
print(df.median())
print("\nMIN:")
print(df.min())
print("\nMAX:")
print(df.max())
print("\nMODE:")
print(df.mode())
print("\nCOUNT (Non-NaN values):")
print(df.count())
print("\nSTANDARD DEVIATION:")
print(df.std())
print("\nVARIANCE:")
print(df.var())
print("\n25th Quantile:")
print(df.quantile(0.25))
print("\nCORRELATION:")
print(df.corr())
print("\nDESCRIBE:")
print(df.describe())
print("\nINDEX OF MAX VALUE:")
print(df.idxmax())
print("\nINDEX OF MIN VALUE:")
print(df.idxmin())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create a sample DataFrame matching the variables from your dataset
df = pd.DataFrame({
    "Sales": [100, 150, 200, 120, 180],
    "Marketing": [80, 130, 170, 90, 150],
    "Ad_Spend": [50, 85, 110, 60, 95],
    "Website_Views": [500, 750, 900, 550, 800],
    "Customer_Satisfaction": [4.5, 4.1, 3.8, 4.6, 3.9]
})

# 2. Calculate the correlation matrix
correlation_matrix = df.corr()

# 3. Plot the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix, 
    annot=True,          # Displays the correlation numbers inside the squares
    cmap="coolwarm",     # Uses the blue (negative) to red (positive) color scale
    vmin=-1,             # Locks the bottom of the color scale to -1.0
    vmax=1,              # Locks the top of the color scale to 1.0
    fmt=".2f",           # Formats the numbers to 2 decimal places
    linewidths=0.5       # Adds a thin line between cells to make it look like a grid
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a simple dataset of random numbers (5 rows, 5 columns)
df = pd.DataFrame(np.random.random((5, 5)), columns=["A", "B", "C", "D", "E"])

# 2. Draw the heatmap
sns.heatmap(df)

# 3. Show the plot
plt.show()
```

```python
# e.g., coolwarm, Blues, viridis
# vmin=-1,             # Locks the bottom of the color scale to -1.0
#     vmax=1,              # Locks the top of the color scale to 1.0
#     fmt=".2f",           # Formats the numbers to 2 decimal places
#     linewidths=0.5       # Adds a thin line between cells to make it look like a grid
sns.heatmap(df, annot=True, cmap="viridis")
# annotate  = annot
plt.show()
```
