```python
# visulization : matplotlib , seaborn 
```

```python
x = [1,2,3,4,5]
y = [10,20,15,30,25]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y, color = "red", marker = "*", markerfacecolor = "green", markeredgecolor = "green")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
names = ["A", "B", "C", "D"]
marks = [99, 65, 90, 70]
plt.bar(names, marks)
plt.xlabel("Names")
plt.ylabel("Marks")
plt.show()
```

```python
x = [10, 20, 30, 40 , 50]
y = [15, 25, 20 , 45, 35]

plt.scatter(x,y)
plt.show()
```

```python
marks = [45, 50, 55, 60, 65, 70, 72, 75, 80, 85, 90 , 95]
plt.hist(marks)
plt.show()
```

```python
subjects = ["Python", "C", "Java", "Sql"]
hours = [10, 5, 8 , 7]
plt.pie(hours, labels = subjects, autopct= "%0.2f%%")
plt.show()
```

```python
import seaborn as sns
```

```python
data = sns.load_dataset("tips")
data
sns.scatterplot(data = data, x = "total_bill", y = "tip")
plt.show()
```

```python
data = sns.load_dataset("flights")
data
sns.lineplot(data = data , x = "year", y = "passengers")
plt.show()
```

```python
data = sns.load_dataset("tips")

sns.barplot(data = data , x = "day", y = "total_bill")
plt.show()
```

```python
data = sns.load_dataset("tips")
sns.boxplot(data = data , x = "day", y = "total_bill")
plt.show()
```

```python
corr = data.corr(numeric_only = True)
sns.heatmap(corr, annot = True , cmap = "cool")
plt.show()
```
