```python
df.isnull().sum()
```

```python
df["City"].unique()
```

```python
df["City"].mode()[0]
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
int(df["Experience"].mean())
```

```python
df["Experience"] = df["Experience"].fillna(int(df["Experience"].mean()))
```

```python
df["Department"].unique()
```

```python
df["Department"] = df["Department"].fillna("General")
```

```python
df["Name"] = df["Name"].fillna("no_name")
```

```python
df["Age"].mean()
```

```python
df["Age"].median()
```

```python
df["Age"].mode()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].sum()
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["EmployeeID"].count()
```

```python
df.groupby("Department")["Salary"].transform("mean")
```

```python
df["Salary"] = df["Salary"].fillna(df.groupby("Department")["Salary"].transform("mean"))
```
