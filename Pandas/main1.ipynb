```python
# Correlation: Think of correlation as a measure of how perfectly two variables dance together. It strips away the original
# units (like years or rupees) and always returns a clean score between -1.0 and 1.0.
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace'],
    'Age': [25, 30, 35, 40, 25, 30, 50],
    'Salary': [50000, 60000, 70000, 80000, 52000, 61000, 95000]
}
df = pd.DataFrame(data)

# 1. Drop string columns to create a numbers-only DataFrame
# (Assuming 'Occup' is not in this dataset, we just drop 'Name')
df_num = df.drop('Name', axis=1)
print("--- Correlation Matrix ---")
print(df_num.corr())
```

```python
# Covariance (df.cov())
# Covariance is correlation's raw, unpolished older sibling. It answers the same basic question—do these two variables move
# in the same direction?—but it doesn't standardize the scale.

# A positive number means they grow together.

# A negative number means they move in opposite directions.

import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace'],
    'Age': [25, 30, 35, 40, 25, 30, 50],
    'Salary': [50000, 60000, 70000, 80000, 52000, 61000, 95000]
}
df = pd.DataFrame(data)

# 1. Drop string columns to create a numbers-only DataFrame
# (Assuming 'Occup' is not in this dataset, we just drop 'Name')
df_num = df.drop('Name', axis=1)
print("--- Correlation Matrix ---")
print(df_num.cov())
```

```python
# Difference (df.diff())
# This function looks at your data sequentially and asks, "What is the exact gap between this step and the step right before it?"
# If your Salary column is [50000, 60000, 75000], calling .diff() will output [NaN, 10000, 15000]
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace'],
    'Age': [25, 30, 35, 40, 25, 30, 50],
    'Salary': [50000, 60000, 70000, 80000, 52000, 61000, 95000]
}
df = pd.DataFrame(data)
print(df)

# 1. Drop string columns to create a numbers-only DataFrame
# (Assuming 'Occup' is not in this dataset, we just drop 'Name')
df_num = df.drop('Name', axis=1)
print("---  Difference between consecutive rows ---")
print(df_num.diff())
```

```python
# froupby and aggregation code

import pandas as pd

df = pd.DataFrame({
    'City': ['Delhi', 'Delhi', 'Mumbai', 'Mumbai'],
    'Item': ['Laptop', 'Phone', 'Laptop', 'Phone'],
    'Sales': [50000, 20000, 60000, 25000],
    'Units': [1, 2, 2, 3]
})
print(df.groupby('City')['Sales'].sum())
```

```python
# aggregation 

print(df.groupby('City').agg({
    'Sales': 'sum',
    'Units': 'mean'
}))
```

```python
# While groupby outputs data as a single vertical list, pivot_table reshapes your data into a two-dimensional grid (Excel-style) with headers
# on both the rows and the columns.

# Question: Build a grid showing Sales by City (rows) vs Item (columns)
print(df.pivot_table(index='City', columns='Item', values='Sales', aggfunc='sum'))
```

```python
import pandas as pd

# --- 1. CONCAT (Stacking Data) ---
class_a = pd.DataFrame({"Name": ["Alice", "Bob"], "Math": [85, 90]})
class_b = pd.DataFrame({"Name": ["Charlie", "David"], "Math": [75, 88]})

# Stacks Class B directly under Class A
all_students = pd.concat([class_a, class_b], ignore_index=True)


# --- 2. MERGE (Matching by Column) ---
emails = pd.DataFrame({"Name": ["Alice", "Bob"], "Email": ["a@mail.com", "b@mail.com"]})

# Stitches the tables together where the "Name" column matches
merged_df = pd.merge(class_a, emails, on="Name")


# --- 3. JOIN (Matching by Index) ---
# First, let's make the "Name" column the actual Index (row label) for both tables
class_a_indexed = class_a.set_index("Name")
emails_indexed = emails.set_index("Name")

# Now we can use join() to stitch them together using that Index
joined_df = class_a_indexed.join(emails_indexed)

```
