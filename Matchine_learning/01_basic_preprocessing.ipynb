```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name" : ["user1","user2", "user3"],
    "Gender" : ["Male", "Female", "Male"],
    "Age" : [21,23, 22]
})
df
```

```python
# label encoding  : converts categorical values into numerical 
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
df["Gender_status"] = encoder.fit_transform(df["Gender"])
```

```python
df
```

```python
# one-hot encoding :
# One-Hot Encoding is a data preprocessing technique that converts categorical (text) data into separate columns containing 0 and 1
df["City"] = ["Mumbai", "Surat", "Ahmedabad"]
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False)

status = encoder.fit_transform(df[["City"]])
status
```

```python
status
```

```python
# status — contains the encoded numerical values produced by your encoder.

# columns=encoder.get_feature_names_out(["City"]) — gets the output column names from the encoder, such as City_Ahmedabad, City_Mumbai, and City_Surat.

# index=df.index — uses the same row index as your original DataFrame.
```

```python
# We use this code in ML preprocessing to add One-Hot Encoded columns to our original dataset
encoded_df = pd.DataFrame(
    status,
    columns=encoder.get_feature_names_out(["City"]),
    index=df.index
)

df = pd.concat([df, encoded_df], axis=1)

df
```

```python
# minmaxscaler

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Dataset
df = pd.DataFrame({
    "Age": [20, 30, 40, 50],
    "Salary": [10000, 20000, 30000, 40000]
})

print("Original Data:")
print(df)

# Apply MinMaxScaler
scaler = MinMaxScaler()

df[["Age", "Salary"]] = scaler.fit_transform(
    df[["Age", "Salary"]]
)

print("\nScaled Data:")
print(df.round(2))
```

```python

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Dataset
df = pd.DataFrame({
    "Age": [20, 30, 40, 50],
    "Salary": [10000, 20000, 30000, 40000]
})

print("Original Data:")
print(df)

# Apply StandardScaler
scaler = StandardScaler()

df[["Age", "Salary"]] = scaler.fit_transform(
    df[["Age", "Salary"]]
)

print("\nStandardized Data:")
print(df.round(2))
```

```python
# give me re mydata after stndard scaler 

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Original data
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
})

# Create scaler
scaler = StandardScaler()

# Scale the data
scaled_data = scaler.fit_transform(df[["Age", "Salary"]])

# Convert scaled data into DataFrame
scaled_df = pd.DataFrame(
    scaled_data,
    columns=["Age", "Salary"]
)

print("Scaled Data:")
print(scaled_df.round(2))

# Get original values back
original_data = scaler.inverse_transform(scaled_data)

# Convert original values into DataFrame
original_df = pd.DataFrame(
    original_data,
    columns=["Age", "Salary"]
)

print("\nOriginal Age and Salary:")
print(original_df)
```
