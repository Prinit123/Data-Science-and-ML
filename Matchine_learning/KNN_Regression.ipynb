```python
# why we use stratify=y

from sklearn.model_selection import train_test_split
import pandas as pd

X = pd.DataFrame({"StudyHours": range(100)})
y = pd.Series(["Pass"] * 80 + ["Fail"] * 20)

# Without stratify
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

print("Without stratify:")
print(y_test.value_counts())

# With stratify
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

print("\nWith stratify:")
print(y_test.value_counts())
```

```python
# create knn model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train,y_train)

y_pred = model.predict(X_test)

print("Actual:",y_test.tolist())
print("predit:-",y_pred.tolist())
print("Accuracy:",accuracy_score(y_test,y_pred))

```

```python
new_student = pd.DataFrame({"StudyHours":[4]})
prediction = model.predict(new_student)

print("new student Prediction: ",prediction[0])
```
